I have tried creating a synthetic data using numpy and pandas without using actual data.

```python
np.random.seed(42) # fixes randomvalues , everytime we run gives same random values

num_samples = 200  # no of samples

age = np.random.randint(18, 25, num_samples) #only use integers btw 18 and 25
study_hours = np.random.normal(5, 2, num_samples)   # mean=5, std=2 #uses normal distribution as most students study max 5 hrs
attendance = np.random.uniform(60, 100, num_samples) #uniform - every values has equal change to get attendance btw 60 - 100

# Target variable (realistic relation)
marks = (
    study_hours * 8 #(each hour we get 8 marks)
     + attendance * 0.4 #(attendacne also effetcs the amrks .ex: 80% -> 32 marks)
    + np.random.normal(0, 5, num_samples) #Real-life is not perfect, so mean-0 , std = 5 gives some noise
)

df = pd.DataFrame({
    "Age": age,
    "Study_Hours": study_hours,
    "Attendance": attendance,
    "Marks": marks.round(2)
})

df.head()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Study_Hours"], df["Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df = pd.read_csv("data.csv")
df.head()
print(df.to_string())
```

```python
type = (df['Age'].value_counts())
```

```python
import matplotlib.pyplot as plt
#Bar Graph = Categorical Data + Aggregation
# Group data - bar graph doesnt work properly with continuous data
avg_marks_by_age = df.groupby("Age")["Marks"].mean()

# Plot bar graph
plt.bar(avg_marks_by_age.index, avg_marks_by_age.values)

plt.xlabel("Age")
plt.ylabel("Average Marks")
plt.title("Average Marks by Age")

plt.show()
```

```python
plt.bar(avg_marks_by_age.index, avg_marks_by_age.values,color='lightgreen')
#zip - pairs one age to one average mark.
for x, y in zip(avg_marks_by_age.index, avg_marks_by_age.values):
    plt.text(x, y + 0.5, f"{y:.1f}", ha='center') #placing of text and formatting decimals

plt.xlabel("Age")
plt.ylabel("Average Marks")
plt.title("Average Marks by Age")
plt.show()
```

```python
#cut function changes numbers to ranges(bins)
df["Attendance_Level"] = pd.cut(
    df["Attendance"],
    bins=[60, 70, 80, 90, 100],
    labels=["60-70", "70-80", "80-90", "90-100"]
    #By default left side is exclusive & right side is inclusive
)
```

```python
avg_marks_att = df.groupby("Attendance_Level")["Marks"].mean()

plt.bar(avg_marks_att.index, avg_marks_att.values)

plt.xlabel("Attendance Range (%)")
plt.ylabel("Average Marks")
plt.title("Average Marks by Attendance Range")

plt.show()
```

Histogram = distribution + frequency, not categories

```python
import matplotlib.pyplot as plt
import numpy as np

# Create histogram
counts, bins, patches = plt.hist(df["Marks"], bins=10,color='lightgreen',edgecolor='black')

# Set labels and title
plt.xlabel("Marks")
plt.ylabel("Number of Students")

# Add values on top of bars
for count, patch in zip(counts, patches):
    plt.text(
        patch.get_x() + patch.get_width() / 2,
        count,
        int(count),
        ha='center',
        va='bottom'
    )

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")

plt.show()
```

```python
plt.scatter(df["Study_Hours"], df["Marks"])

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks (Scatter Plot)")

plt.show()
```

Scatter + regression

```python
# Scatter plot
plt.scatter(df["Study_Hours"], df["Marks"])

# Regression line
m, c = np.polyfit(df["Study_Hours"], df["Marks"], 1)
plt.plot(df["Study_Hours"], m * df["Study_Hours"] + c)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks with Regression Line")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(24, 18),constrained_layout=True)  # overall figure size
```

```python
plt.subplot(2, 2, 1)

counts, bins, patches = plt.hist(df["Marks"], bins=10)
plt.tight_layout
# add values on histogram bars
for count, patch in zip(counts, patches):
    plt.text(
        patch.get_x() + patch.get_width() / 2,
        count,
        int(count),
        ha='center',
        va='bottom'
    )

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Students")

#sCATTER
plt.tight_layout
plt.subplot(2, 2, 2)

plt.scatter(df["Study_Hours"], df["Marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
#bARGRAPH
plt.tight_layout
plt.subplot(2, 2, 3)

avg_marks_by_age = df.groupby("Age")["Marks"].mean()

plt.bar(avg_marks_by_age.index, avg_marks_by_age.values)

plt.title("Avg Marks by Age")
plt.xlabel("Age")
plt.ylabel("Average Marks")
#PIECHART
plt.tight_layout
plt.subplot(2, 2, 4)

attendance_counts = df["Attendance_Level"].value_counts()

plt.pie(
    attendance_counts.values,
    labels=attendance_counts.index,
    autopct="%1.1f%%"
)

plt.title("Attendance Level Distribution")
plt.tight_layout
plt.show()
```

```python
plt.subplot(2, 2, 2)

# Scatter plot (actual data)
plt.scatter(df["Study_Hours"], df["Marks"], label="Data Points")

# Regression line calculation
m, c = np.polyfit(df["Study_Hours"], df["Marks"], 1)
#1 - degree #m - slope #c - intercept
# Plot regression line
plt.plot(
    df["Study_Hours"],
    m * df["Study_Hours"] + c,
    label="Regression Line"
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend()
#Differentiates:
#Data points
#Regression line
```
